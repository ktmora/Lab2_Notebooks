```python
import numpy as np
import matplotlib.pyplot as plt

# ----------------------------
# 7.1: Combine + Smooth spectra
# ----------------------------

def moving_smooth(y, w, method="mean"):
    """
    Smooth 1D array y with a window of width w (channels).
    method: "mean" or "median"
    Uses reflection padding so edges don't dip.
    """
    w = int(w)
    if w < 1:
        return y.copy()
    if w % 2 == 0:
        w += 1  # enforce odd window for symmetric smoothing

    pad = w // 2
    ypad = np.pad(y, (pad, pad), mode="reflect")

    if method == "mean":
        kernel = np.ones(w, dtype=float) / w
        return np.convolve(ypad, kernel, mode="valid")

    elif method == "median":
        # sliding window median (vectorized via strides)
        from numpy.lib.stride_tricks import sliding_window_view
        windows = sliding_window_view(ypad, window_shape=w)
        return np.median(windows, axis=-1)

    else:
        raise ValueError("method must be 'mean' or 'median'")


def combined_power_spectrum(npz_file, fs, M=None, combine="mean"):
    """
    Returns (f_if, P_combined) where:
      f_if is baseband IF frequency axis (Hz)
      P_combined is combined linear power spectrum (NOT dB)
    combine: "mean" or "median" across captures (robust to RFI)
    """
    arr = np.load(npz_file)["arr_1"]  # (captures, N, 2)
    ncap, N, _ = arr.shape
    if M is None:
        M = ncap
    M = min(M, ncap)

    window = np.hanning(N)
    W2 = np.mean(window**2)

    # Compute power spectrum for each capture
    P_stack = np.empty((M, N), dtype=np.float64)

    for m in range(M):
        x = arr[m, :, 0] + 1j * arr[m, :, 1]
        x = x - np.mean(x)  # remove DC
        X = np.fft.fftshift(np.fft.fft(x * window))
        P_stack[m] = (np.abs(X) ** 2) / (N**2 * W2)

    # Combine across time/captures
    if combine == "mean":
        P_comb = np.mean(P_stack, axis=0)
    elif combine == "median":
        P_comb = np.median(P_stack, axis=0)
    else:
        raise ValueError("combine must be 'mean' or 'median'")

    f_if = np.fft.fftshift(np.fft.fftfreq(N, d=1/fs))  # Hz
    return f_if, P_comb


# ----------------------------
# USER SETTINGS
# ----------------------------
fs = 2.9e6
npz_file = "on2.npz"       # change to "off2.npz" as needed
M = 1000                   # number of captures to combine

combine_method = "mean"    # "mean" or "median" across captures
smooth_method  = "mean"    # "mean" or "median" across channels

# Rule of thumb: HI line not narrower than ~1 km/s -> ~4.74 kHz at 1420 MHz
nu0 = 1420.4058e6
dv_kms = 1.0
df_target_hz = nu0 * (dv_kms * 1e3) / 299792458.0  # ~4.74 kHz

# Convert target smoothing width (Hz) -> channels
# You can scale this (e.g., 1x, 2x) depending on how noisy your spectrum is.
# Keep it modest to preserve line shape.
smooth_width_hz = df_target_hz  # try 1x; if still noisy, try 2x
# ----------------------------

f_if, P = combined_power_spectrum(npz_file, fs, M=M, combine=combine_method)

df_chan = np.abs(f_if[1] - f_if[0])        # Hz/channel
w_chan = int(np.round(smooth_width_hz / df_chan))
w_chan = max(w_chan, 1)

P_smooth = moving_smooth(P, w=w_chan, method=smooth_method)

# Plot (linear + dB)
P_db = 10*np.log10(P + 1e-20)
P_smooth_db = 10*np.log10(P_smooth + 1e-20)

plt.figure(figsize=(12, 7))
plt.plot(f_if/1e3, P_db, alpha=0.35, label=f"Combined ({combine_method}) unsmoothed")
plt.plot(f_if/1e3, P_smooth_db, linewidth=2.0,
         label=f"Smoothed ({smooth_method}), window={w_chan} ch (~{w_chan*df_chan/1e3:.2f} kHz)")
plt.xlabel("Baseband IF Frequency Offset (kHz)")
plt.ylabel("Power (dB)")
plt.title(f"{npz_file}: Combine={combine_method}, Smooth={smooth_method}")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

print(f"Data: {npz_file}")
print(f"df per channel = {df_chan:.3f} Hz")
print(f"Smoothing window = {w_chan} channels  (~{w_chan*df_chan:.1f} Hz)")
print(f"1 km/s target width ~ {df_target_hz:.1f} Hz")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from numpy.lib.stride_tricks import sliding_window_view

# ----------------------------
# Helpers
# ----------------------------
def moving_smooth(y, w, method="mean"):
    w = int(w)
    if w < 2:
        return y.copy()
    if w % 2 == 0:
        w += 1
    pad = w // 2
    ypad = np.pad(y, (pad, pad), mode="reflect")

    if method == "mean":
        kernel = np.ones(w, dtype=float) / w
        return np.convolve(ypad, kernel, mode="valid")
    elif method == "median":
        windows = sliding_window_view(ypad, window_shape=w)
        return np.median(windows, axis=-1)
    else:
        raise ValueError("method must be 'mean' or 'median'")

def combined_power_spectrum(npz_file, fs, M=None, combine="mean"):
    arr = np.load(npz_file)["arr_1"]  # (captures, N, 2)
    ncap, N, _ = arr.shape
    if M is None:
        M = ncap
    M = min(M, ncap)

    window = np.hanning(N)
    W2 = np.mean(window**2)

    P_stack = np.empty((M, N), dtype=np.float64)
    for m in range(M):
        x = arr[m, :, 0] + 1j * arr[m, :, 1]
        x = x - np.mean(x)
        X = np.fft.fftshift(np.fft.fft(x * window))
        P_stack[m] = (np.abs(X)**2) / (N**2 * W2)

    if combine == "mean":
        P_comb = np.mean(P_stack, axis=0)
    elif combine == "median":
        P_comb = np.median(P_stack, axis=0)
    else:
        raise ValueError("combine must be 'mean' or 'median'")

    f_if = np.fft.fftshift(np.fft.fftfreq(N, d=1/fs))
    return f_if, P_comb

def median_baseline(y, w):
    """Wide median filter baseline estimator."""
    w = int(w)
    if w < 3:
        return y.copy()
    if w % 2 == 0:
        w += 1
    pad = w // 2
    ypad = np.pad(y, (pad, pad), mode="reflect")
    windows = sliding_window_view(ypad, window_shape=w)
    return np.median(windows, axis=-1)

# ----------------------------
# Files + settings
# ----------------------------
fs = 2.9e6

on_file  = "on2.npz"
off_file = "off2.npz"

cal_file  = "bodytest1.npz"
cold_file = "nobodytest1.npz"

M_onoff = 1000
M_cal   = 1000

combine_method = "mean"   # also TRY "median" after this works
smooth_method  = "mean"

T_sys_cal = 300.0

# smoothing width: 1 km/s ~ 4.74 kHz
nu0 = 1420.4058e6
df_target_hz = nu0 * (1e3 / 299792458.0)

# ----------------------------
# 1) Compute G
# ----------------------------
f_cal, s_cal = combined_power_spectrum(cal_file, fs, M=M_cal, combine=combine_method)
f_cold, s_cold = combined_power_spectrum(cold_file, fs, M=M_cal, combine=combine_method)

if not np.allclose(f_cal, f_cold):
    raise RuntimeError("cal and cold frequency axes do not match.")

df_chan = np.abs(f_cal[1] - f_cal[0])
w_chan = max(1, int(np.round(df_target_hz / df_chan)))

s_cal_s  = moving_smooth(s_cal,  w=w_chan, method=smooth_method)
s_cold_s = moving_smooth(s_cold, w=w_chan, method=smooth_method)

sum_scold = np.sum(s_cold_s)
sum_diff  = np.sum(s_cal_s - s_cold_s)
if sum_diff <= 0:
    raise RuntimeError("sum(s_cal - s_cold) <= 0")

G = T_sys_cal * (sum_scold / sum_diff)

print(f"GAIN G = {G:.3f} K")
print(f"df per channel = {df_chan:.3f} Hz")
print(f"smoothing window = {w_chan} ch (~{w_chan*df_chan/1e3:.2f} kHz)")

# ----------------------------
# 2) r_line = s_on/s_off
# ----------------------------
f_on, s_on = combined_power_spectrum(on_file, fs, M=M_onoff, combine=combine_method)
f_off, s_off = combined_power_spectrum(off_file, fs, M=M_onoff, combine=combine_method)

if not np.allclose(f_on, f_off):
    raise RuntimeError("ON and OFF frequency axes do not match.")

s_on_s  = moving_smooth(s_on,  w=w_chan, method=smooth_method)
s_off_s = moving_smooth(s_off, w=w_chan, method=smooth_method)

r_line = s_on_s / (s_off_s + 1e-30)

# ----------------------------
# 3) Robust baseline removal
# ----------------------------

# A) Throw away band edges (FIR rolloff): keep central 70% of band
fmax = np.max(np.abs(f_on))
edge_frac = 0.70
good = (np.abs(f_on) < edge_frac * fmax)

f_g = f_on[good]
r_g = r_line[good]

# B) Median-filter baseline window (VERY wide: tens of kHz to a few 100 kHz)
# Try ~200 kHz baseline window:
baseline_width_hz = 200e3
w_base = max(11, int(np.round(baseline_width_hz / df_chan)))
if w_base % 2 == 0:
    w_base += 1

baseline = median_baseline(r_g, w_base)

# Flatten ratio and compute HI excess temperature
r_flat = r_g / baseline
T_HI = (r_flat - 1.0) * G

print("\n--- After robust baseline removal ---")
print(f"T_HI min = {np.min(T_HI):.3f} K")
print(f"T_HI max = {np.max(T_HI):.3f} K")
print(f"T_HI median = {np.median(T_HI):.3f} K")
print(f"baseline window = {w_base} ch (~{w_base*df_chan/1e3:.1f} kHz)")

# ----------------------------
# Plot
# ----------------------------
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 9), sharex=True)

ax1.plot(f_g/1e3, r_g, linewidth=1.2, label="r_line (cropped)")
ax1.plot(f_g/1e3, baseline, linestyle="--", linewidth=1.5, label="median baseline")
ax1.axhline(1.0, linestyle=":", linewidth=1.0)
ax1.set_ylabel("r_line")
ax1.set_title("Robust baseline removal (median filter) + edge cropping")
ax1.grid(True)
ax1.legend()

ax2.plot(f_g/1e3, T_HI, linewidth=1.4)
ax2.axhline(0.0, linestyle=":", linewidth=1.0)
ax2.set_xlabel("Baseband IF Frequency Offset (kHz)")
ax2.set_ylabel(r"$T_{\mathrm{HI}} = (r_{\mathrm{flat}}-1)G$ (K)")
ax2.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

from astropy.time import Time
from astropy.coordinates import SkyCoord, EarthLocation, AltAz
import astropy.units as u

import ugradio

# ------------------------------------------------------------
# Inputs you already have:
#   T_HI : HI excess temperature (K)
#   nu_rf (Hz) from your earlier plotting cell
# ------------------------------------------------------------

# ----------------------------
# Instrument settings (yours)
# ----------------------------
fLO   = 1420.395e6       # Hz
nu0   = 1420.4058e6      # Hz
c_kms = 299792.458       # km/s

# ----------------------------
# REDIFINE f_g (DO NOT CHANGE ANYTHING ELSE)
# ----------------------------
f_g = nu_rf - fLO   # cropped baseband IF frequency axis (Hz), consistent with your previous code

# ----------------------------
# Observation time + location
# ----------------------------
utc_time = "2026-02-18 19:31:00"
t = Time(utc_time, scale="utc")

location = EarthLocation(lat=37.873199*u.deg, lon=-122.2573*u.deg, height=0*u.m)

print("UTC:", utc_time)
print("Julian Date:", f"{t.jd:.9f}")

# ------------------------------------------------------------
# 1) Compute RA/Dec of ZENITH at that time/location
# ------------------------------------------------------------
# Zenith in AltAz is alt=90 deg (az is irrelevant at the zenith)
zenith_altaz = SkyCoord(alt=90*u.deg, az=0*u.deg, frame=AltAz(obstime=t, location=location))

# Transform to ICRS (RA/Dec)
zenith_icrs = zenith_altaz.icrs
ra_deg  = zenith_icrs.ra.deg
dec_deg = zenith_icrs.dec.deg

print("\nZenith pointing (computed from time+location):")
print(f"RA  = {ra_deg:.9f} deg")
print(f"Dec = {dec_deg:.9f} deg")

# ------------------------------------------------------------
# 2) Observer-frame velocity axis (matches your earlier code)
# ------------------------------------------------------------
nu_rf = fLO + f_g
delta_nu = nu_rf - nu0
v_kms = -c_kms * (delta_nu / nu0)

# ------------------------------------------------------------
# 3) ugradio LSR correction  (CORRECT ARG ORDER)
# ------------------------------------------------------------
jd = t.jd

try:
    v_proj = ugradio.doppler.get_projected_velocity(ra_deg, dec_deg, jd)
except TypeError:
    v_proj = ugradio.doppler.get_projected_velocity(
        ra_deg, dec_deg, jd,
        obs_lat=37.873199, obs_lon=-122.2573
    )

# If ugradio returns m/s, convert to km/s
if np.abs(v_proj) > 500:
    print("\nNOTE: ugradio returned a large value; assuming m/s and converting to km/s.")
    v_proj = v_proj / 1000.0

print(f"\nugradio projected velocity correction = {v_proj:.6f} km/s")

# ------------------------------------------------------------
# 4) Apply correction + plot
# ------------------------------------------------------------
# Convention: usually v_LSR = v_obs + v_proj, but some installs flip sign.
# We'll compute both and plot both so you can pick the correct one.
v_lsr_add = v_kms + v_proj
v_lsr_sub = v_kms - v_proj

plt.figure(figsize=(12,6))
plt.plot(v_kms, T_HI, label="Observer frame", linewidth=1.4)
plt.plot(v_lsr_add, T_HI, label=f"LSR (v_obs + {v_proj:.2f})", linewidth=1.4)
#plt.plot(v_lsr_sub, T_HI, label=f"LSR (v_obs - {v_proj:.2f})", linewidth=1.2, alpha=0.7)
plt.axhline(0.0, linestyle=":", linewidth=1.0)
plt.axvline(0.0, linestyle="--", linewidth=1.0)
plt.xlabel("Velocity (km/s)")
plt.ylabel(r"$T_{\mathrm{HI}}$ (K)")
plt.title("Observer vs LSR (ugradio) using RA/Dec computed from zenith")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 5) Peak comparison (to decide sign)
# ------------------------------------------------------------
imax = np.argmax(T_HI)
print("\n--- Peak velocity comparison at max(T_HI) ---")
print(f"Observer peak v       = {v_kms[imax]:.2f} km/s")
print(f"LSR peak (add)        = {v_lsr_add[imax]:.2f} km/s")
print(f"LSR peak (subtract)   = {v_lsr_sub[imax]:.2f} km/s")
```

```python
import numpy as np
import matplotlib.pyplot as plt

importdata = np.load("cassy_data.npz")
indexdata  = importdata["arr_1"]          # shape ~ (captures, samples, 2)

sampling_rate = 2.9e6                     # <-- set correctly
fLO = 1420.395e6                          # <-- real center frequency

# Choose how many captures to average
M = min(1000, indexdata.shape[0])           # try 50 to start

# Use the same N for every FFT
N = indexdata.shape[1]

# Window (and a normalization so power scale is consistent)
window = np.hanning(N)
W2 = np.mean(window**2)                   # window power normalization

P_accum = np.zeros(N, dtype=np.float64)

for m in range(M):
    data = indexdata[m]                   # (N,2)
    x = data[:,0] + 1j*data[:,1]          # complex IQ

    # Remove DC offset (helps a lot)
    x = x - np.mean(x)

    # FFT
    X = np.fft.fftshift(np.fft.fft(x * window))
    P = (np.abs(X)**2) / (N**2 * W2)      # linear power, normalized

    P_accum += P

P_avg = P_accum / M

# Frequency axis (baseband)
f = np.fft.fftshift(np.fft.fftfreq(N, d=1/sampling_rate))  # Hz

# Convert to dB only at the end (add epsilon to avoid -inf)
P_avg_db = 10*np.log10(P_avg + 1e-20)



# Plotting 

fig, (ax1, ax2) = plt.subplots(nrows=2, ncols=1, figsize=(10,8), sharex=True)

# First Plot
ax1.plot(f/1e3, P_avg, color="teal")
ax1.set_xlabel("Baseband IF Frequency Offset (kHz)")
ax1.set_ylabel("Power")
ax1.set_title(f"Averaged Baseband Power Spectrum (Blocks={M})")
ax1.grid(True)

# Second Plot 
ax2.plot(f/1e3, P_avg_db, color="darkorchid")
ax2.set_xlabel("Baseband IF Frequency Offset (kHz)")
ax2.set_ylabel("Power (dB)")
#plt.xlim(-500,200)
ax2.set_title(f"Averaged Baseband Power Spectrum (Blocks={M})")
ax2.grid(True)

plt.tight_layout()
plt.show()
```
